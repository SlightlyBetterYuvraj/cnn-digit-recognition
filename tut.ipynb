```python
import pandas as pd
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784', version=1, as_frame=True)

df = mnist.data.iloc[:60000].copy()
df['label'] = mnist.target.iloc[:60000].astype(int)

df_test = mnist.data.iloc[60000:].copy()
df_test['label'] = mnist.target.iloc[60000:].astype(int)
```

```python
df.head()
```

```python
xtrain = df.drop("label", axis=1).values
ytrain = df["label"].values
xtest = df_test.drop("label", axis=1).values
ytest = df_test["label"].values
```

```python
xtrain = xtrain.astype("float32") / 255.0
xtest = xtest.astype("float32") / 255.0
```

```python
xtrainimg = xtrain.reshape(-1, 28, 28)
xtestimg = xtest.reshape(-1, 28, 28)
```

```python
from tensorflow.keras.utils import to_categorical

ytraincat = to_categorical(ytrain, 10)
ytestcat = to_categorical(ytest, 10)
```

```python
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras import Sequential


percep = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(10, activation="softmax")
])
```

```python
percep.compile(optimizer="sgd", loss="categorical_crossentropy", metrics=["accuracy"])
```

```python
histpercep = percep.fit(xtrainimg, ytraincat, epochs=5, batch_size=32, validation_data=(xtestimg, ytestcat), verbose=1)
```

#ANN

```python
ann = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation="relu"),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax"),
])
```

```python
ann.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
histpercep = ann.fit(xtrainimg, ytraincat, epochs=5, batch_size=32, validation_data=(xtestimg, ytestcat), verbose=1)
```

```python
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dropout

cnn = Sequential([
    Conv2D(32, kernel_size=(3,3), activation="relu", input_shape=(28,28,1)),
    MaxPooling2D(pool_size=(2,2)),
    Conv2D(64, kernel_size=(3,3), activation="relu"),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.5),
    Dense(10, activation="softmax")
])
```

```python
xtraincnn = xtrain.reshape(-1, 28, 28, 1)
xtestcnn = xtest.reshape(-1, 28, 28, 1)
```

```python
cnn.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
histpercep = cnn.fit(xtraincnn, ytraincat, epochs=5, batch_size=32, validation_data=(xtestcnn, ytestcat), verbose=1)
```
